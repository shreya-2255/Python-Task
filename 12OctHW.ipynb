## Finding topper of each subject marks.

```python
db = {
    1: [78, 97, 86],
    2: [97, 94, 92],
    3: [78, 90, 87]
}

# Maths
topper_roll = 1
topper_marks = db[1][0]

for k, v in db.items():
    if v[0] > topper_marks:
        topper_roll = k
        topper_marks = v[0]

print("Maths topper: Roll No.", topper_roll, "Marks:", topper_marks)

# Physics
topper_roll = 1
topper_marks = db[1][1]

for k, v in db.items():
    if v[1] > topper_marks:
        topper_roll = k
        topper_marks = v[1]

print("Physics topper: Roll No.", topper_roll, "Marks:", topper_marks)

# Biology
topper_roll = 1
topper_marks = db[1][2]

for k, v in db.items():
    if v[2] > topper_marks:
        topper_roll = k
        topper_marks = v[2]

print("Biology topper: Roll No.", topper_roll, "Marks:", topper_marks)
```
