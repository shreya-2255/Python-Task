## Q.3 WAP to check whether string is palindrom or not.

```python
p = "nayan"
orignal = p
p_reverse = ""
for ch in p:
    p_reverse = p_reverse + p[-1]
    p = p[:len(p)-1]

if orignal == p_reverse:
    print("Palindrom")
else:
    print("Not Palindrom")
```
