## Q.4 WAP to check whether string is palindrom or not.

```python
p = "nayan"
original = p
p_reverse = ""
for ch in p:
    p_reverse = p_reverse + p[-1]
    p = p[:len(p)-1]

if original == p_reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

## Q.5 WAP to count the number of occurrences of the character 'i' in the given string.

```python
s = "indivisibility"
count = 0
for ch in s:
    if ch == 'i':
        count += 1

print(count)
```

## Q.7 WAP to remove all spaces from a given string without using the replace() method.

```python
s = "Java By Kiran"
new_s = ""

for ch in s:
    if ch != " ":
        new_s += ch

print(new_s)
```
