## Write a program to check whether given number is prime or not.

```python
num = int(input("Enter the number: "))

count = 0

if num <= 1:
    print(num, "is not a prime number")
else:
    for i in range(2, num):
        if num % i == 0:
            count += 1

    if count == 0:
        print(num, "is a prime number")
    else:
        print(num, "is not a prime number")
```
