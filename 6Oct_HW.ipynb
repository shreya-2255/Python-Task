## WAP to print names of student who are absent today.

```python
Total_student = ["Raj", "Amit","Jay", "Payal", "Priya", "Rahul", "Pavan", "Kiran", "Prachi"]
Present_attendance = ("Raj", "Jay", "Payal", "Priya", "Pavan", "Kiran")
absent_student = []

for stu in Total_student:
    if stu not in Present_attendance:
        absent_student.append(stu)

print(absent_student)
```

## Methods in List data type
 1. append
 2. clear
 3. copy
 4. count
 5. extend
 6. index
 7. insert
 8. pop
 9. rempve
 10. reverse
 11. sort
 

```python
students = ['Raj','Neetu','Manoj','Jonh','Priya']
print(type(students))
print(dir(students))
print(students)
```

```python
## 1. append()
students.append('Geetu')
print(students)
```

```python
## 2. insert()
students.insert(4, 'Arya')
print(students)
```

```python
## 3. extend()
new_students = ['Ragav', 'Ronny']
students.extend(new_students)
print(students)
```

```python
## 4. remove()
students.remove('Jonh')
print(students)
```

```python
## 5. pop()
students.pop()
print(students)
students.pop(3)
print(students)
```

```python
## 6. clear()
students.clear()
print(students)
```

```python
## 7. index()
students = ['Raj','Neetu','Manoj','Jonh','Priya']
print(students)
print(students.index('Jonh'))
```

```python
## 8. count()
numbers = [10, 20, 30, 20, 60, 20, 80, 40, 20]
print(numbers.count(20))
```

```python
## 9. sort()
numbers = [60, 20, 40, 80, 10, 30]
numbers.sort()
print(numbers)
## for desending order
numbers.sort(reverse=True)
print(numbers)
```

```python
## 10. reverse ()
numbers=[10, 70, 50, 90, 40,30]
numbers.reverse()
print(numbers)
```

```python
## 11. copy()
students = ['Raj','Neetu','Manoj','Jonh','Priya']
new_students = students.copy()
print(new_students)
```

```python
## adding functions on copied list

new_students.append('Jay')
print(new_students)
print(students)
```
