## Write a program to get unique element list from above list without changing data's position.

```python
li= [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]
uni_li = []

for i in li:
    if i not in uni_li:
        uni_li.append(i)

print(uni_li)
```

## Write a program to find frequency of every element from above input list in dict.

```python
li = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]
freq_ele = {}

for i in li:
    if i in freq_ele:
        freq_ele[i] += 1
    else:
        freq_ele[i] = 1

print(freq_ele)
```

## Write a program to find frequency of every character from above input string in dict.

```python
str = "Python Programming"

freq_ch = {}

for ch in str:
    if ch in freq_ch:
        freq_ch[ch] += 1
    else:
        freq_ch[ch] = 1

print(freq_ch)
```
