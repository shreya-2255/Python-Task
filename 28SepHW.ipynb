## Arithmetic Operators
1. Int
2. Float
3. Complex
4. Boolean
5. Complex


```python
a = 25
b = 5

print(a + b)   
print(a - b)   
print(a * b)   
print(a / b)   
print(a % b)   
print(a // b)  
```

```python
a = 10.5
b = 2.5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a // b)
```

```python
a = 2 + 3j
b = 1 + 2j

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
a = "Hello"
b = "World"

print(a + b)      
print(a * 3)      
```

## Comparision Operator

```python
a = 15
b = 25

print(a > b)     # False
print(a < b)     # True
print(a >= b)    # False
print(a <= b)    # True
print(a == b)    # False
print(a != b)    # True
```

```python
a = 10.5
b = 5.5

print(a > b)     # True
print(a < b)     # False
print(a == b)    # False
```
