## Arithmetic Operators
1. Int
2. Float
3. Complex
4. Boolean
5. Complex


```python
a = 10
b = 3

print(a + b)   
print(a - b)   
print(a * b)   
print(a / b)   
print(a % b)   
print(a // b)  
```

```python
a = 10.5
b = 2.5

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a % b)
print(a // b)
print(a ** b)
```

```python
a = 2 + 3j
b = 1 + 2j

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
a = "Hello"
b = "World"

print(a + b)      
print(a * 3)      
```

```python
a = True
b = False

print(a + b)   
print(a * b)  
print(a - b)  
```

## Logical Operator

```python
a = 10
b = 20

print(a < b and b > 15)   
print(a > b or b > 15)   
print(not(a < b))         
```

```python
a = 10.5
b = 5.5

print(a > b and b > 5)    
print(a < b or b > 5)     
print(not(a > b))         
```
